## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Official FlyRank starter dataset
DATA_URL = (
    "https://huggingface.co/datasets/FlyRank/internship-starter/"
    "resolve/main/content_refresh_anonymized.csv"
)

df = pd.read_csv(DATA_URL)

# Key evidence for the research lane
corr = df["search_volume"].corr(df["impressions_90d"])
avg_ctr = df["ctr"].mean()
avg_position = df["avg_position"].mean()

print(f"Dataset size: {len(df):,} rows")
print(f"Search volume vs impressions correlation: {corr:.4f}")
print(f"Average CTR: {avg_ctr:.2f}%")
print(f"Average position: {avg_position:.2f}")
```
